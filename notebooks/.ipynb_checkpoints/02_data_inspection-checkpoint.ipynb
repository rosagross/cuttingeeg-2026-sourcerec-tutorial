## Data Inspection
- Plot EEG data
- compute noise covariance
- compute data covariance


```python
# imports
import mne
import os
import matplotlib.pyplot as plt 
import numpy as np

# data paths
subID = 18
data_path = "data"
subject_path = os.path.join(f"sub-0{subID}", "ses-mecha", "eeg")
epoch_file = f"sub-0{subID}_ses-mecha_task-NT_epo.fif"
epochs_path = os.path.join("..", data_path, subject_path, epoch_file)
```

### EEG Data
We load and inspect data from one participant.

```python
# load data
epochs = mne.read_epochs(epochs_path, preload=True)


# inspect EEG data object
epochs
```

```python
spectrum = epochs.compute_psd()
spectrum.plot_topomap()
```

```python
epochs.plot_image(picks="O2", combine="mean")
```

```python
evoked = epochs.average()
evoked.plot()
```

```python
unperceived = epochs["response_decoded=='No'"].average()
perceived = epochs["response_decoded=='Yes'"].average()
```

```python
plt.plot(np.squeeze(unperceived.get_data(picks="O2")), label='perceived')
plt.plot(np.squeeze(perceived.get_data(picks="O2")), label='unperceived')
plt.legend()
```
