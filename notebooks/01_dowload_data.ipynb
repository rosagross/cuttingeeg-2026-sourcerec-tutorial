# 1. Data Download
The data used for this workshop is publicly available at Zenodo, an all-purpose repository.
The url is already filled in, just run the following cells and the download will start automatically into our data folder.
There are exactly three files:
- the epoch file (containing the EEG data)
- the forward model (which is precomputed for this tutorial)
- the morph-to-avergage matrix (for plotting the reconstructed data on an average brain surface)

```python
from pathlib import Path
import requests

DATA_DIR = Path("../data/sub-024/ses-mecha/eeg")
DATA_DIR.mkdir(parents=True, exist_ok=True)

files = {
    "sub-024_ses-mecha_task-NT_epo.fif":
        "https://zenodo.org/records/22692782/files/sub-024_ses-mecha_task-NT_epo.fif",

    "sub-024_ses-mecha_task-NT_fwd.fif":
        "https://zenodo.org/records/22692782/files/sub-024_ses-mecha_task-NT_fwd.fif",

    "sub-024_ses-mecha_task-NT_morph.h5":
        "https://zenodo.org/records/22692782/files/sub-024_ses-mecha_task-NT_morph.h5",
}
```

```python
for filename, url in files.items():
    destination = DATA_DIR / filename
    
    if destination.exists():
        print(f"✓ {filename} already exists")
        continue
    
    print(f"Downloading {filename}...")
    
    response = requests.get(url)
    response.raise_for_status()
    
    destination.write_bytes(response.content)
    
    size_mb = destination.stat().st_size / 1024**2

    print(f"Downloaded successfully: {destination}")
    print(f"File size: {size_mb:.1f} MB")
```

### Check if data was downloaded into the correct folder

```python
datapath = DATA_DIR
print("Download folder:", datapath)
print(datapath.resolve(), "\n")
print("Path exists:", datapath.exists(), "\n")
print("Files in the folder:\n", list(datapath.iterdir()))
```
